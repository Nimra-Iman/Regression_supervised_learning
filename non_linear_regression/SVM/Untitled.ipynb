```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dataset=pd.read_csv('Position_Salaries.csv')
dataset=dataset.drop(columns=['Position'], axis=1)
dataset.head()
```

```python
x=dataset[['Level']]
y=dataset['Salary']
```

```python
from sklearn.preprocessing import StandardScaler
ss=StandardScaler()
x=pd.DataFrame(ss.fit_transform(x), columns=x.columns)
```

```python
plt.figure(figsize=(5,4))
sns.scatterplot(x='Level', y='Salary', data=dataset)
plt.show()
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test= train_test_split(x,y, test_size=0.3, random_state=42)
```

```python
# 'linear', 'poly', 'rbf', 'sigmoid', 'precomputed'--> precomputed for those data jis m ziada 0 and 1 hon
# , yani hm n agr onehotencoding ki hui h to precomputed ko bhi use kia ja skta h
from sklearn.svm import SVR
# Evaluate the model
svm = SVR(kernel='poly')
svm.fit(x_train, y_train)

```

```python
# Evaluate the model
# Evaluate the model
accuracy_train = svm.score(x_train, y_train)
accuracy_test = svm.score(x_test, y_test)
accuracy_train*100, accuracy_test*100
```

```python
svm.predict([[1]])
```

```python
plt.figure(figsize=(5,4))
sns.scatterplot(x='Level', y='Salary', data=dataset)
sns.lineplot(x=dataset['Level'], y=svm.predict(x),data=dataset, color="red")
plt.show()
```
