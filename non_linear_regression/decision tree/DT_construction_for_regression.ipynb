```python
When we have many input columns (features) in a decision tree regression problem, we do not use only one column. Instead, for every node, we test all columns one by one against the output column. For example, if we have:

1. Column 1
2. Column 2
3. Column 3

then we will try different split points for:

* Column 1 vs output
* Column 2 vs output
* Column 3 vs output

For every possible split, we calculate the error using the mean approach (mean of output values in each region). Then we compare all errors and select the split that gives the minimum error. The feature and split point with the least error become the node split.

For example, if Column 3 gives the minimum error and the best split is:
“Column 3 < 1700”

then the root node becomes based on Column 3.

After that, the dataset is divided into two parts:

* rows where Column 3 < 1700
* rows where Column 3 ≥ 1700

Now for the second split, we do NOT use the entire dataset again. Instead, we only use the subset of rows that reached that branch. Inside that subset, we again test all columns:

* Column 1
* Column 2
* Column 3

and again calculate all possible errors to find the best next split. The same feature can also be reused multiple times in different parts of the tree if it provides the best split.
```
